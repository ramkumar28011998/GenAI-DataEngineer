#### AUTO MODEL CLASS METHOD

####Using Transformer LLMs for Sentiment Analysis
* Here we don't need to mention explicitly for any class details

```python
comment_1 = "I absolutely love this product. It works perfectly and exceeded my expectations"
comment_2 = "This service was very disappointing and a complete waste of money"
comment_3 = "The package arrived yesterday and was placed on the table"
```

Q1: Perform sentiment analysis on given customer comments.
1. Do not use pipeline function
2. Use Auto Classes
3. Apply distilbert-base-uncased-finetuned-sst-2-english model from huggingface hub

Step 1 - Create a tokenizer and prepare input tokens

```python
from transformers import AutoTokenizer

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
```

##### In the above step, tokenizer class will be automatically chosen based on given model type

How to prepare the inputs using the tokenizer?

##### Below step is straight solution to perform all the pre processing stages in a single go

```python
inputs = tokenizer([comment_1, comment_2, comment_3], padding=True, return_tensors="pt")
print(inputs)
```

Step 2 - Create an use a model to generate raw outputs

```python
from transformers import AutoModelForSequenceClassification

model = AutoModelForSequenceClassification.from_pretrained(checkpoint)
raw_outputs = model(inputs.input_ids, inputs.attention_mask)
print(raw_outputs)
```

Step 3 - Convert raw outputs to probabilities

```python
import torch

predictions = torch.nn.functional.softmax(raw_outputs.logits, dim=1)
print(predictions)
```

Step 4 - Format and display the results

```python
    import pandas as pd

    pd_result_df = pd.DataFrame(predictions.tolist()) \
                        .rename({0: model.config.id2label[0].title(),
                                1: model.config.id2label[1].title()}, axis=1)

    pdf_comments = pd.DataFrame([comment_1, comment_2, comment_3], columns=['Comments']) 
    pd_final_result_df = pdf_comments.join(pd_result_df)
    display(pd_final_result_df)
    
```

##### MODEL CLASS METHOD
* Here we need to explicitly mention the class details based on Model type

```python
from transformers import DistilBertTokenizer
checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = DistilBertTokenizer.from_pretrained(checkpoint)
```

```python
## HERE WE IMPORTED THE DISTIL BERT MODEL TOKENIZER
```

```python
inputs = tokenizer([comment_1, comment_2, comment_3], padding=True, return_tensors="pt")
print(inputs)
```

```python
from transformers import DistilBertForSequenceClassification
model = DistilBertForSequenceClassification.from_pretrained(checkpoint)
outputs = model(inputs.input_ids, inputs.attention_mask)
print(outputs)
```

```python
import torch
predictions = torch.nn.functional.softmax(outputs.logits, dim=1) #apply to each row
print(predictions)
```

```python
import pandas as pd
pdf_results = pd.DataFrame(predictions.tolist()) \
                .rename({0: model.config.id2label[0].title(), 
                         1: model.config.id2label[1].title()}, 
                        axis=1)
pdf_comments = pd.DataFrame([comment_1, comment_2, comment_3], columns=['Comment']) 
pd_result_df = pdf_comments.join(pdf_results)
display(pd_result_df)               
```

## USING PIPELINE OBJECT TO HANDLE THE PRE PROCESSING STAGES

Step 1 - Create the pipeline

```python
from transformers import pipeline
classifier = pipeline(task="sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
```

#### STRAIGHT A WAY TO ACCESS THE MODEL AND ASKING THE QUESTIONS

```python
results = classifier([comment_1, comment_2, comment_3])
print(results)
```

#### PIPELINE OBJECT HELPS US TO BUILD PRE PROCESSING STAGE

Step 2 - Extract a tokenizer and prepare input tokens

```python
tokenizer = classifier.tokenizer
inputs = tokenizer([comment_1, comment_2, comment_3], padding=True, return_tensors="pt")
print(inputs)
```

Step 3 - Extract your model from the pipeline

```python
model = classifier.model
outputs = model(inputs.input_ids, inputs.attention_mask)
print(outputs)
```

Step 4 - Convert raw outputs to probabilities

```python
import torch
predictions = torch.nn.functional.softmax(outputs.logits, dim=1) #apply to each row
print(predictions)
```

Step 5 - Format and display the results

```python
import pandas as pd
pdf_results = pd.DataFrame(predictions.tolist()) \
                .rename({0: model.config.id2label[0].title(), 
                         1: model.config.id2label[1].title()}, 
                        axis=1)
pdf_comments = pd.DataFrame([comment_1, comment_2, comment_3], columns=['Comment']) 
pd_result_df = pdf_comments.join(pdf_results)
display(pd_result_df)               
```
